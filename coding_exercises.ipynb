##### 3. Local variable

Local variables can be useful as they store information within the function only, without affecting the rest of the code, unlike global variables that are assigned to the whole program. Using local variables makes function more independent and predictable.



```python
def calculate_area():
    width = 5
    height = 3
    area = width * height
    print(area)

calculate_area()
```

In this example **width**, **height**, and **area** are local variables. 

Important thing to remember is that the local variables do not ovewrite global variables. This is useful because different parts of a program can use the same variable names without interfering with each other.

```python
name = "Polina"

def say_hello():
    name = "Alex"
    print(name)

say_hello()
print(name)
```
