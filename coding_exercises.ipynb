Microsoft Copilot introduced functions in a following way:

**Function** creates a named set of instructions that can be used later. Functions allow to reuse code without rewriting it, organise programs into smaller separate pieces, make code easier to read and maintain, avoid repeating the same instructions many time.

```python
def say_hello():
```

##### My Additional Prompt: 
> Is there anything else I should learn about function on the beginner level? 

I wanted to see what are the other basics that would be useful to learn before moving to more complicated concepts. 

Microsoft Copilot introduced me to advancements such as parameters, arguments, return values, local variables, functions invoked by functions.

##### 1. Parameters and arguments
Parameters allow function to work with different information. Basically, parameters are variables that exist inside the function. 


```python
def say_hello(name):
    print(f"Hello, {name}!")

say_hello("Polina")
say_hello("Alex")
say_hello("Mark")
```

**name** is the parameter
**"Polina** is the argument. 

A parameter is a placeholder for the value that needs to be put inside the parameter. And an argument is the actual value. Agument can be changed depending on what we want to be displayed.

##### 2. Return values

Functions can also produce the results. 

```python
def add(a, b):
    return a + b
result = add(2, 3)
print(result)
```

The value of the **return** can be almost anything: a text, a calculation, true or flase, list, multiple values. 

***True or Flase Example***

```python
def is_adult(age):
    return age >= 18
print(is_adult(20))
print(is_adult(15))
```

***Calculation Example***

```python
def square(number):
    return number ** 2
print(square(5))
```

##### 3. Local variable

Local variables can be useful as they store information within the function only, without affecting the rest of the code, unlike global variables that are assigned to the whole program. Using local variables makes function more independent and predictable.



```python
def calculate_area():
    width = 5
    height = 3
    area = width * height
    print(area)

calculate_area()
```

In this example **width**, **height**, and **area** are local variables. 

Important thing to remember is that the local variables do not ovewrite global variables. This is useful because different parts of a program can use the same variable names without interfering with each other.

```python
name = "Polina"

def say_hello():
    name = "Alex"
    print(name)

say_hello()
print(name)
```

##### 4. Functions that call other functions

Larger multifucntional programs are composed from smaller code pices like this. 

```python
def say_hello():
    print("Hello!")

def welcome():
    say_hello()
    print("Welcome to Python!")

welcome ()
```

- **for** is the key word that. starts a loop, which tells the program to repear instructions for each item in a sequence.

- **range(5)** tells Python to count through 5 values.

Now we can actually see the value changing each time the loop repeats.
##### My Additional Prompt:

The underscore is a special convention that means that the code receives a value, but i doesn't matter what value it is.

#### My additional promt. 

#### *Lists + loops*: 
I decided to explore how it would be possible to greet everyone with a **loop**.

> Show how to greet everyone with a loop

```python
names = ["Polina", "Alex", "Sam"]

for name in names:
    print(f"Hello, {name}!")
```

- **name** in the loop is a variable that holds temporarily one item from the list at a time. Python creates and updates this variable automatically as the loop runs.
-**names** is a variable that stores all the values from the list. 

This code allows to take each individual name from the collection of names.

#### *Function + list + loop*

Useful to perform the same operation multiple times.

```python
def say_hello():
    names = ["Polina", "Alex", "Sam"]

    for name in names:
        print(f"Hello, {name}!")

say_hello()
```

It is also possible to pass list as an argument.

```python
def say_hello(names):
    for name in names:
        print(f"Hello, {name}!")

friends = ["Polina", "Alex", "Sam"]

say_hello(friends)
```

In this example **friend** contains the **list**. And then Python gives that list to the parameter **names**.

#### *Function + list for calculating the total*

Functions and lists are often used together for numbers. 

```python
def add_numbers(numbers):
    total = 0

    for number in numbers:
        total = total + number

    return total

scores = [10, 20, 30]

print(add_numbers(scores))
```

### Dictionaries
#### Initial promt: 
> Could you suggest a modification of the Python code for "Hello,world!" to demonstrate the concept of dictionaries to a complete beginner in programming? Please explain the code step-by-step highlighting any key aspects. 

```python
person = {
    "name": "Polina"
}

print(f"Hello, {person['name']}!")
```

This creates a variable called person.

The value stored in person is a dictionary.

A dictionary stores data as key-value pairs:

|Key   |Value   |
|----- |--------|
|"name"|"Polina"|

- dictionaries use curly braces **{}**
- each entry has a **key** and a **value**
- the key and value are separated by a colon **:**
- multiple entries are separated by **commas**

##### Adding more information to the dictionary.

Dictionaries are useful when they store several pieces of related information. 

```python
person = {
    "name": "Polina",
    "age": 20,
    "city": "Glasgow"
}

print(f"Hello, {person['name']}!")
print(person["age"])
print(person["city"])
```

This example of a dictionary contains more information about the person including name, age, and city. 

|Key   |Value    |
|------|---------|
|"name"|"Polina" |
|"age" |20       |
|"city"|"Glasgow"|

Once the information is in the dictionary we can easily retrieve it when we need.

### Arrays
#### Initial promt: 
> Could you suggest a modification of the Python code for "Hello,world!" to demonstrate the concept of arrays to a complete beginner in programming? Please explain the code step-by-step highlighting any key aspects. 

```python
from array import array

numbers = array('i', [1, 2, 3])

for number in numbers:
    print(f"Hello, world! Number: {number}")
```

- an **array** stores multiple values in one variable.
- every element has an index.
- **arrays** usually contain the same type of data.
- **array** commonly stores numbers, not text; **lists** are more appropriate for words/names
- a **loop** can process every element automatically. 

Arrays let to store many related values in one variable and process them efficiently.
