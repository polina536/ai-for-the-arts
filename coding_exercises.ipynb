### Functions
##### Initial Prompt:
> Could you suggest a modification of the Python code for "Hello, world!" to demonstrate the concept of functions to a complete beginner in programming? Pleas explain the code step-by-step highlighting any key aspects. 

Microsoft Copilot introduced functions in a following way:

```python
def say_hello():
    print("Hello, world!")

say_hello()
```

**Function** creates a named set of instructions that can be used later. Functions allow to reuse code without rewriting it, organise programs into smaller separate pieces, make code easier to read and maintain, avoid repeating the same instructions many time. 

##### 1. Defining a function

```python
def say_hello():
```

**Function** creates set of instructions to be reused later. 

- def is a Python keyword that means "define a function."
- say_hello is the name of the function.
- The parentheses () are required, even though this function doesn't need any information passed to it yet.
- The colon : tells Python that the function's instructions are about to begin.

##### 2. The function body

```python
 print("Hello, world!")
```

This line tells Python what to do when the function is executed. In this way Python creates a function but has not run it yet. 

##### 3. Calling the function

```python
say_hello()
```

- Writing the function's name followed by () tells Python to run the function.
- This is called **calling** or **invoking** a function.

##### My Additional Prompt:

##### 1. Parameters and arguments
Parameters allow function to work with different information. Basically, parameters are variables that exist inside the function. 


```python
def say_hello(name):
    print(f"Hello, {name}!")

say_hello("Polina")
say_hello("Alex")
say_hello("Mark")
```

**name** is the parameter
**"Polina** is the argument. 

A parameter is a placeholder for the value that needs to be put inside the parameter. And an argument is the actual value. Agument can be changed depending on what we want to be displayed.

##### 2. Return values

Functions can also produce the results. 

```python
def add(a, b):
    return a + b
result = add(2, 3)
print(result)
```

The value of the **return** can be almost anything: a text, a calculation, true or flase, list, multiple values. 

***True or Flase Example***

```python
def is_adult(age):
    return age >= 18
print(is_adult(20))
print(is_adult(15))
```

***Calculation Example***

```python
def square(number):
    return number ** 2
print(square(5))
```

##### 3. Local variable

Local variables can be useful as they store information within the function only, without affecting the rest of the code, unlike global variables that are assigned to the whole program. Using local variables makes function more independent and predictable.



```python
def calculate_area():
    width = 5
    height = 3
    area = width * height
    print(area)

calculate_area()
```

In this example **width**, **height**, and **area** are local variables. 

Important thing to remember is that the local variables do not ovewrite global variables. This is useful because different parts of a program can use the same variable names without interfering with each other.

```python
name = "Polina"

def say_hello():
    name = "Alex"
    print(name)

say_hello()
print(name)
```

##### 4. Functions that call other functions

Larger multifucntional programs are composed from smaller code pices like this. 

```python
def say_hello():
    print("Hello!")

def welcome():
    say_hello()
    print("Welcome to Python!")

welcome ()
```

### Loops
#### Initial promt: 
> Could you suggest a modification of the Python code for "Hello world!" to demonstrate the concept of loops to a complete beginner in progamming? Please explain the code step-by-step highlighting any key aspects. 

```python
for i in range(5):
    print("Hello, world!")
```

- **for** is the key word that. starts a loop, which tells the program to repear instructions for each item in a sequence. 
- **i** is a loop variable, it takes different values. 
- **range(5)** tells Python to count through 5 values. 
- **print()** is the statement that we are going to run 5 times, 4 spaces before **print** show that it belongs to the loop. 

In this example we have put the loop variable, but haven't actually used it. 

```python
for i in range(5):
    print(i, "Hello, world!")
```

Now we can actually see the value changing each time the loop repeats.
I have asked Microsoft Copilot whether it is possible to run a loop without a loop variable. 
However, it is important to note that the loop always needs something between **for** and **in** to hold each value produced by range().



```python
for _ in range(5):
    print("Hello, world!")
```

The underscore is a special convention that means that the code receives a value, but i doesn't matter what value it is.
